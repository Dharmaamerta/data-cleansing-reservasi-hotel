**Membaca data kotor reservasi hotel**

```python
file_path = '/content/drive/MyDrive/Itn/Data minning/Data kotor reservasi hotel.xlsx'
df = pd.read_excel(file_path, engine='openpyxl')
```

**Mencetak data dari google drive**

```python
print(df)
```

**Melihat Struktur Dataset**

```python
print("Ukuran Data set:",df.shape)
print("\nNama kolom:",df.columns.tolist())
print("\nTipe data:")
print(df.dtypes)
```

**Mengecek Data kosong**

```python
print("Cek data kosong:")
print(df.isnull().sum())
```

**Membersihkan Spasi**

```python
for col in df.columns:
    if df[col].dtype == 'object':
        df[col] = df[col].str.strip()
```

**Standarisasi Tipe Data**

```python
print(df["ID Reservasi"].unique())
```

Standarisasi Umur

```python
df["Umur"] = df["Umur"].astype(str).str.extract(r'(\d+)')[0]
df["Umur"] = pd.to_numeric(df["Umur"], errors="coerce")

print(df["Umur"].unique())
```

Standarisasi Jenis kelamin


```python
df['Jenis Kelamin'] = df['Jenis Kelamin'].str.strip().str.lower()
df['Jenis Kelamin'] = df['Jenis Kelamin'].replace({
    'l': 'Laki-laki',
    'laki-laki': 'Laki-laki',
    'p': 'Perempuan',
    'perempuan': 'Perempuan'
})
print(df["Jenis Kelamin"].unique())
```

Standarisasi Tipe Kamar

```python
df['Tipe Kamar'] = df['Tipe Kamar'].str.strip().str.lower()
df['Tipe Kamar'] = df['Tipe Kamar'].replace({
    'delux': 'Deluxe',
    'deluxe': 'Deluxe',
    'suite': 'Suite',
    'standard': 'Standard'
})

print(df['Tipe Kamar'].unique())
```

Standarisasi Metode Pembayaran



```python
df['Metode Pembayaran'] = df['Metode Pembayaran'].str.strip().str.lower()
df['Metode Pembayaran'] = df['Metode Pembayaran'].replace({
    'transfer': 'Transfer',
    'TRANSFER': 'Transfer',
    'tunai':'Tunai',
    'qris':'Qris',
    'QRIS':'Qris',
    'suite': 'Suite',
    'standard': 'Standard'
})

print(df['Metode Pembayaran'].unique())
```

Standarisasi Status Reservasi

```python
df['Status Reservasi'] = df['Status Reservasi'].str.strip().str.lower()

df['Status Reservasi'] = df['Status Reservasi'].replace({
    'selesai': 'Selesai',
    'dibatalkan': 'Dibatalkan'
})
print(df['Status Reservasi'].unique())
```
